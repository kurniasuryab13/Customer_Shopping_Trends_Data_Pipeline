# Transform

```python
df.dtypes # mengecek tipe data pada setiap column
```

```python
df['Customer ID'] = df['Customer ID'].astype('str') # mengubah tipe data menjadi string pada column Customer ID
```

```python
df['Subscription Status'] = df['Subscription Status'].replace({'Yes': True, 'No': False}).astype(bool)
df['Discount Applied'] = df['Discount Applied'].replace({'Yes': True, 'No': False}).astype(bool)
df['Promo Code Used'] = df['Promo Code Used'].replace({'Yes': True, 'No': False}).astype(bool)
# mengubah nilai 'Yes' menjadi True dan 'No' menjadi False pada column Subscription Status, Discount Applied, dan Promo Code Used, kemudian mengubah tipe data menjadi boolean (bool) pada ketiga column tersebut.
```

# Great Expectations

```python
from great_expectations.data_context import FileDataContext # import lib GX yang diperlukan.

context = FileDataContext.create(project_root_dir='./') # untuk membuat GX context yang berbentuk folder bernama gx.
```

```python
datasource_name = 'shop_trend_nov_9_min_1' # memberikan nama pada Datasource.
datasource = context.sources.add_pandas(datasource_name)

asset_name = 'shop_trend_nov' # memberikan nama pada asset.
path_to_data = 'data_cleaned.csv' # mengambil data dari file csv bernama data_cleaned.csv, kemudian disimpan pada variabel path_to_data.
asset = datasource.add_csv_asset(asset_name, filepath_or_buffer=path_to_data) # syntax untuk meng-load/register path data 


batch_request = asset.build_batch_request() # untuk membaca datanya
```

```python

expectation_suite_name = 'expectation-shoptrends-dataset' # memberikan nama pada suite dengan nama expectation-shoptrends-dataset.
context.add_or_update_expectation_suite(expectation_suite_name) # membuat expectation suite.

validator = context.get_validator( 
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
) # untuk menyambungkan suite ke data yang telah dibaca/akan digunakan.

validator.head() # menampilkan data (5 teratas).
```

```python
# Expectation 1 : Column `Customer ID` harus unique

validator.expect_column_values_to_be_unique('Customer ID')
```

```python
# Expectation 2 : Column `Age` harus berada di antara 1 sampai 110

validator.expect_column_values_to_be_between(
    column='Age', min_value=1, max_value=110
)
```

```python
# Expectation 3 : Column `Payment Method` harus mengandung salah satu dari 6 hal berikut:
# Credit Card
# Bank Transfer
# Cash
# PayPal
# Venmo
# Debit Card

validator.expect_column_values_to_be_in_set('Payment Method', ['Credit Card', 'Bank Transfer', 'Cash', 'PayPal', 'Venmo',
       'Debit Card']) 
```

```python
# Expectation 4 : Column `Discount Applied` harus dalam bentuk tipe boolean.

validator.expect_column_values_to_be_of_type(
    "Discount Applied",
    "bool"
)
```

```python
# Expectation 5 : Column `Review Rating` harus memiliki nilai median antara 1 sampai 5.

validator.expect_column_median_to_be_between(
    column="Review Rating",
    min_value=1,
    max_value=5
)
```

```python
# Expectation 6 : Column `Purchase Amount (USD)` harus memiliki nilai mean antara 50 sampai 150.

validator.expect_column_mean_to_be_between(
    column="Purchase Amount (USD)",
    min_value=50,
    max_value=150
)
```

```python
# Expectation 7 : Column `Category` harus memiliki pasangan nilai yang benar dengan column `Item Purchased`.

validator.expect_column_pair_values_to_be_in_set(
        'Category',
        'Item Purchased',
        [
            ('Clothing','Blouse'),
            ('Clothing','Dress'),
            ('Clothing','Hoodie'),
            ('Clothing','Jeans'),
            ('Clothing','Pants'),
            ('Clothing','Shirt'),
            ('Clothing','Shorts'),
            ('Clothing','Skirt'),            
            ('Clothing','Socks'),
            ('Clothing','Sweater'),
            ('Clothing','T-shirt'),
            ('Accessories','Backpack'),            
            ('Accessories','Belt'),
            ('Accessories','Gloves'),
            ('Accessories','Handbag'),
            ('Accessories','Hat'),            
            ('Accessories','Jewelry'),
            ('Accessories','Scarf'),
            ('Accessories','Sunglasses'),
            ('Accessories','yellow'),            
            ('Footwear','Boots'),
            ('Footwear','Sandals'),
            ('Footwear','Sneakers'),
            ('Footwear','Shoes'),            
            ('Outerwear','Coat'),
            ('Outerwear','Jacket'),  
        ]
)
```

```python
validator.save_expectation_suite(discard_failed_expectations=False)
# save validator tersebut ke dalam suite.
```

```python
checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_1',
    validator = validator,
)
# membuat checkpoint.
```

```python
checkpoint_result = checkpoint_1.run() # menjalankan checkpoint 1.
```
